```python
from pyspark.sql.functions import *

catalog = "bike_data"
bronze_schema = "bronze"

tables_to_analyze = [
    "crm_cust_info",
    "crm_prd_info",
    "crm_sales_details",
    "erp_cust_az12",
    "erp_loc_a101",
    "erp_px_cat_g1v2"
]

for table_name in tables_to_analyze:
    print("\n" + "=" * 80)
    print(f"DATA QUALITY ANALYSIS: {table_name}")
    print("=" * 80)
    
    df = spark.table(f"{catalog}.{bronze_schema}.{table_name}")
    
    print(f"Total rows: {df.count()}")
    print(f"Total columns: {len(df.columns)}")
    
    print("\nSchema:")
    df.printSchema()
    
    print("\nNull values per column:")
    null_check = df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns])
    display(null_check)
    
    print("\nFirst 5 rows:")
    display(df.limit(5))
```
