```python
import time

catalog = "bike_data"
bronze_schema = "bronze"

print("=" * 80)
print("ORCHESTRATION: Bronze Layer")
print("=" * 80)

print("\nRunning: bronze_ingestion")
start_time = time.time()

try:
    dbutils.notebook.run("/Workspace/Users/e51888557@gmail.com/bike_data_lakehouse/bronze/bronze_ingestion", timeout_seconds=3600)
    elapsed_time = time.time() - start_time
    print(f"\n bronze_ingestion completed in {elapsed_time:.2f} seconds")
except Exception as e:
    print(f"\n❌ bronze_ingestion failed: {str(e)}")
    raise

# Verify Bronze tables
print("\n" + "=" * 80)
print("VERIFICATION: Bronze Tables")
print("=" * 80)

bronze_tables = spark.sql(f"SHOW TABLES IN {catalog}.{bronze_schema}").collect()
print(f"\nBronze tables created: {len(bronze_tables)}")

for table in bronze_tables:
    table_name = table.tableName
    count = spark.sql(f"SELECT COUNT(*) FROM {catalog}.{bronze_schema}.{table_name}").collect()[0][0]
    print(f"   {table_name}: {count:,} rows")

print("\n" + "=" * 80)
print("Bronze layer orchestration completed successfully")
print("=" * 80)
```
