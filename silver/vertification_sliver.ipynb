```python
from pyspark.sql.functions import *

catalog = "bike_data"
silver_schema = "silver"_

silver_tables = [
    "customers",
    "products", 
    "sales",
    "erp_customers",
    "locations",
    "categories"
]

for table_name in silver_tables:
    print(f"\n{table_name}:")
    df = spark.table(f"{catalog}.{silver_schema}.{table_name}")
    df.printSchema()
    print(f"Sample:")
    display(df.limit(1))
```
