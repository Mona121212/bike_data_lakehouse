```python
catalog = "bike_data"

print("=" * 80)
print("FINAL PIPELINE VERIFICATION")
print("=" * 80)

# Check all layers
print("\nBronze Layer:")
bronze_count = spark.sql(f"SELECT COUNT(*) FROM {catalog}.bronze.crm_cust_info").collect()[0][0]
print(f"  crm_cust_info: {bronze_count:,} rows")

print("\nSilver Layer:")
silver_count = spark.sql(f"SELECT COUNT(*) FROM {catalog}.silver.customers").collect()[0][0]
print(f"  customers: {silver_count:,} rows")

print("\nGold Layer:")
gold_count = spark.sql(f"SELECT COUNT(*) FROM {catalog}.gold.fact_sales").collect()[0][0]
print(f"  fact_sales: {gold_count:,} rows")

# Sample query
print("\n" + "=" * 80)
print("SAMPLE BI QUERY")
print("=" * 80)

result = spark.sql(f"""
    SELECT 
        c.cst_firstname,
        c.cst_lastname,
        COUNT(f.sls_ord_num) as orders,
        SUM(f.sls_sales) as total_sales
    FROM {catalog}.gold.fact_sales f
    JOIN {catalog}.gold.dim_customers c ON f.sls_cust_id = c.cst_id
    GROUP BY c.cst_firstname, c.cst_lastname
    ORDER BY total_sales DESC
    LIMIT 5
""")

display(result)

print("\n" + "=" * 80)
print("✅ Pipeline verification complete!")
print("=" * 80)
```
